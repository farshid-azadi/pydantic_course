> **Master the fundamentals of working with dates, times, time zones, and durations in Python.**

<br>

---

### 📅 Date  
Represents a calendar date: **year, month, and day**

### 🕐 Time  
Represents a time of day: **hour, minute, second, and microsecond**

### 🗓️ Datetime  
Combines both **date** and **time** into one complete timestamp.

### 🌍 Time Zone  
Defines the geographical context of a time and prevents ambiguity across regions.

### ⏱️ Timedelta  
Represents a duration or difference between two dates/times.

````python
from datetime import datetime, timezone
from zoneinfo import ZoneInfo

utc_now = datetime.now(timezone.utc)
tehran_time = utc_now.astimezone(ZoneInfo("Asia/Tehran"))
print("the tehran time is now", tehran_time)
````
