# Numerical Field Constraints

```python
from pydantic import BaseModel, Field, ValidationError
```

For numerical constraints, we have these arguments available:
- `gt` -> `>`
- `ge` -> `>=`
- `lt` -> `<`
- `le` -> `<=`
- `multiple_of` -> number must a multiple of this value

Their usage is very straightforward:

```python
class Model(BaseModel):
    number: int = Field(gt=1, lt=21, multiple_of=2)
```

```python
m = Model(number=4)
m
```

```python
try:
    Model(number=23)
except ValidationError as e:
    print(e.errors()[0]["msg"])
```

```python
from pydantic import PositiveInt


class Model(BaseModel):
    number: float = Field(gt=0)


Model.model_fields
```

```python
class Model(BaseModel):
    number: PositiveInt


Model.model_fields
```

So, `PositiveInt` is really the same as `Field(gt=0)`.

In fact, if you recall we had seen this when we inspected the model fields.

```python
from datetime import datetime, timedelta


class Model(BaseModel):
    day: datetime = Field(gt=datetime.now() + timedelta(hours=1))


datetime.now()
Model(day=datetime.now())
```

```python
from datetime import UTC


def get_time():
    return datetime.now(UTC)


class Model(BaseModel):
    dt: datetime = Field(default_factory=get_time)


Model()
```

```python
Model.model_fields
```

```python
from pydantic import PositiveInt


class Model(BaseModel):
    number: PositiveInt
```
