## 🎬 پرده اول: نمایش “کد کثیف و آسیب‌پذیر” (The Chaos)

```python
class Appointment:
    def __init__(self, service_name: str, price: int, service_list: list):
        if not service_name:
            raise ValueError("نام خدمت خود را بنویسید")
        if not isinstance(price, int) or price <= 0:
            raise ValueError("قیمت باید یک عدد صحیح و مثبت باشد")
        if not isinstance(service_list, list):
            raise ValueError("خدمات باید لیستی از نام ها باشد")
        if len(service_list) == 0:
            raise ValueError("لیست خدمات نباید خالی باشد")
        if len(service_list) > 5:
            raise ValueError("حداکثر اسم پنج خدمت را ذکر کنید")
        for s in service_list:
            if not isinstance(s, str):
                raise ValueError("هر سرویس باید یک رشته متنی باشد")
        self.service_name = service_name
        self.price = price
        self.service_list = service_list


try:
    app1 = Appointment("Mancure", -666, ["Nail Art"])
except ValueError as e:
    print(f"Caught Error:{e}")

try:
    app2 = Appointment("Pedicure", 2200, ["s1", "s2", "s3", "s4", "s5", "s6"])
except ValueError as e:
    print(f"Caught error: {e}")
```

## پرده دوم کد تمیس:CLEAN CODE

```python
from pydantic import BaseModel, PositiveInt, Field
from typing import Annotated

Service_List = Annotated[list[str], Field(min_length=1, max_length=5)]


class Appointment(BaseModel):
    service_name: str = Field(min_length=1)
    price: PositiveInt

    service_list: Service_List


# test
try:
    app1 = Appointment(service_name="Manicure", price=-666, service_list=["Nail Art"])
except Exception as e:
    print(f"پایدانتیک به زیبایی خطا ها را کنترل کرد مثل دستگاه اتو کلاو عمل کرد \n{e}")
try:
    app2 = Appointment(
        service_name="Pedicure",
        price=2200,
        service_list=["s1", "s2", "s3", "s4", "s5"],
    )
except Exception as e:
    print(f"پایدانتیک به زیبایی محدودیت ها را تشخیص داد \n{e}")
```
