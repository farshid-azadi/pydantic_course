## ConLists

```python
from pydantic import BaseModel, ValidationError, PositiveInt
```

```python
class Circle(BaseModel):
    center: tuple[int, int] = (0, 0)
    radius: PositiveInt = 1
```

```python
Circle(center=(1, 2, 3))
```

# or 3d Sphere

```python
class Sphere(BaseModel):
    center: tuple[int, int] | tuple[int, int, int] = (0, 0)
    radius: PositiveInt
```

```python
Sphere(center=(1, 1), radius=10)
```

```python
Sphere(center=(1, 1, 4), radius=10)
```

```python
try:
    Sphere(center=(1, 2, 3, 4))
except ValidationError as ex:
    print(ex)
```

```python
from pydantic import conlist


class Sphere(BaseModel):
    center: conlist(int, min_length=2, max_length=3) = [0, 0]
    radiue: PositiveInt = 1
```

```python
Sphere(center=(1, 2))
```

```python
Sphere(center=(1, 2, 5))
```

```python
try:
    Sphere(center=(1,))
except ValidationError as ex:
    print(ex)
```
